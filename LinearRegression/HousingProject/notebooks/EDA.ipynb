# Project 1. Predicting the housing price

### This code explains the EDA (Exploratory Data Analysis). This includes answering the following questions:
#### Which feature has the biggest effect on the Housing Price?
1. Which feature has largest effect on the housing price?
2. Are there outliers?
3. Are some features highly correlated?
4. Should some features be transformed?
5. Is the dataset balanced?
6. Can we already anticipate problems for Linear Regression?

```python
# load/import packages and libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import fetch_california_housing

housing = fetch_california_housing(as_frame=True)
df = housing.frame
```

```python
# Let's first see how the target variable is distributed
plt.figure(figsize=(8,5))

plt.hist(df["MedHouseVal"], bins=40)

plt.xlabel("Median House Value")
plt.ylabel("Frequency")
plt.title("Distribution of House Prices")

plt.show()
```

```python
# If we want to visualize all the variables.
df.hist(figsize=(15,12), bins=30)

plt.tight_layout()
plt.show()
```

```python
## To see any outlier

plt.figure(figsize=(15,6))

df.boxplot()

plt.xticks(rotation=45)

plt.show()
```

### Correlation matrix. This basically shows how different features are correlated with each other. This is also a way to see which input feature has largest impact on our target variable.

```python
corr = df.corr()

plt.figure(figsize=(10,8))

plt.imshow(corr, cmap="coolwarm", interpolation="nearest")

plt.colorbar()

plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)

plt.title("Correlation Matrix")

plt.show()
```

```python
# Which feature correlates with the house price. This also tells which feature has the biggest impact on the housing price. 
# The larger the correlation the bigger its impact on the house price.
corr["MedHouseVal"].sort_values(ascending=False)
```

## Scatter Plots:
### Correlation matrix shows the summary of the impact of different input features with the target variable. However, to see the effect of a variable with other variables and target variable, we use the scatter plots.

```python
# As an example, let's see how median income is related to the median house value
plt.figure(figsize=(8,6))

plt.scatter(
    df["MedInc"],
    df["MedHouseVal"],
    alpha=0.2
)

plt.xlabel("Median Income")
plt.ylabel("Median House Value")

plt.show()
```

## Geographic Visualisation: This also shows the map of the city/state/locality we are considering.

```python
plt.figure(figsize=(8,8))

plt.scatter(
    df["Longitude"],
    df["Latitude"],
    c=df["MedHouseVal"],
    cmap="viridis",
    s=5
)

plt.colorbar(label="House Value")

plt.xlabel("Longitude")
plt.ylabel("Latitude")

plt.title("California Housing Data")

plt.show()
```

```python
### Pairwise relationships
pd.plotting.scatter_matrix(
   df[
        [
            "MedInc",
            "AveRooms",
            "HouseAge",
            "MedHouseVal"
        ]
    ],
    figsize=(10,10)
)

plt.show()
```
